```python
# Question 1
# Correct solution (without redundancy)
condition1 = df["Winning_Team"]=="Match Tied"
condition2 = df["Balls_Remaining"] <= 3
condition3 = (df["Winning_Margin"] <= 1) & (df["Win_Type"]=="wicket") & (df["Winning_Team"] == "Chasing")
condition4 = (df["Winning_Margin"] <= 5) & (df["Win_Type"]=="run") & (df["Winning_Team"] == "FirstBatting")
df[(condition1) | (condition2) | (condition3) | (condition4)].groupby("Year").count()["Match_Number"]
```

```python
# Question 2
teams = ["Chennai", "Hyderabad", "Mumbai", "Kolkata"]

for team in teams:
    hasTeamWon = (df["Winner"] == team)
    didTeamBatFirst = (df["Team_Batting_First"] == team)

    totalWonMatches = df[hasTeamWon]
    totalWonMatchesBattingFirst = df[hasTeamWon & didTeamBatFirst]

    percent = len(totalWonMatchesBattingFirst) / len(totalWonMatches) * 100
    print(team, "%.2f"%percent)
```
